```python
import matplotlib.pyplot as plt
from torchvision import transforms

fig = plt.figure(figsize=(14, 12))

# Image transformation and data augmentation

# --- Crop transformation ---
# Original image for cropping
img_crop_orig, attr_crop_orig = celebat_train_dataset[0]
ax = fig.add_subplot(2, 5, 1)
ax.set_title('Original for\nCrop', size=15)
ax.imshow(img_crop_orig)

# Cropped image
ax = fig.add_subplot(2, 5, 6)
img_cropped = transforms.functional.crop(img_crop_orig, 50, 20, 128, 128)
ax.set_title('Crop to a \nbounding-box', size=15)
ax.imshow(img_cropped)

# --- Flip transformation ---
# Original image for flipping (using next image from dataset for variety)
img_flip_orig, attr_flip_orig = celebat_train_dataset[1] # Using celebat_train_dataset[1]
ax = fig.add_subplot(2, 5, 2)
ax.set_title('Original for\nFlip', size=15)
ax.imshow(img_flip_orig)

# Flipped image
img_flipped = transforms.functional.hflip(img_flip_orig)
ax = fig.add_subplot(2, 5, 7) # Corrected index from 11 to 7
ax.set_title('Flip horizontally', size=15)
ax.imshow(img_flipped)

#lets modify lighting

## Column 3: adjust contrast
# --- Contrast transformation ---
# Original image for contrast adjustment
img_contrast_orig, attr_contrast_orig = celebat_train_dataset[2]
ax = fig.add_subplot(2, 5, 3)
ax.set_title('Original for\nContrast', size=15)
ax.imshow(img_contrast_orig)

# Adjusted contrast image
ax = fig.add_subplot(2, 5, 8)
# Corrected typo: funtional -> functional
img_contrast = transforms.functional.adjust_contrast(img_contrast_orig, 2)
ax.set_title('Adjust contrast', size=15) # Corrected title
ax.imshow(img_contrast)

# --- Brightness transformation ---
# Original image for brightness adjustment
img_brightness_orig, attr_brightness_orig = celebat_train_dataset[3]
ax = fig.add_subplot(2, 5, 4)
ax.set_title('Original for\nBrightness', size=15)
ax.imshow(img_brightness_orig)

# Adjusted brightness image
img_brightness = transforms.functional.adjust_brightness(img_brightness_orig, 0.5)
ax = fig.add_subplot(2, 5, 9) # Corrected index from 16 to 9
ax.set_title('Adjust brightness', size=15) # Corrected title
ax.imshow(img_brightness) # Corrected to show img_brightness

# --- Center crop and resize transformation ---
# Original image for center crop and resize
img_center_crop_orig, attr_center_crop_orig = celebat_train_dataset[4]
ax = fig.add_subplot(2, 5, 5)
ax.set_title('Original for\nCenter Crop &\nResize', size=15)
ax.imshow(img_center_crop_orig)

# Center cropped and resized image
ax = fig.add_subplot(2, 5, 10)
img_center_crop = transforms.functional.center_crop(img_center_crop_orig, [0.7*218, 0.7*178])
img_resized = transforms.functional.resize(img_center_crop, size=(218, 178))
ax.set_title('Center crop\nand resize', size=15)
ax.imshow(img_resized)

plt.show()
```
